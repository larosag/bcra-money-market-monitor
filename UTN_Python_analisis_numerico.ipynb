En el método de bisección debemos decidir qué mitad del intervalo conservar.

Si

$$
f(a) * f(x_r)<0
$$


la raíz queda entre `a` y `xr`.


````python
def f(x):
    return x**2 - 2

a = 1
b = 2
xr = (a + b) / 2

if f(a) * f(xr) < 0:
    b = xr
else:
    a = xr

print(a, b)
````

# 6. Funciones

Una función permite agrupar instrucciones y reutilizarlas.

Vamos a tener que representar funciones matemáticas y también construir algoritmos que puedan utilizarse con distintas funciones.


La función matemática


$$f(x)=x^2-2$$


puede escribirse:


````python
def f(x):
    return x**2 - 2
````

`x` es el **parámetro** de la función y `return` devuelve el resultado.


````python
print(f(1))
print(f(2))
````

````python
x = -1
if (x<0):
  y = -x
  print(y)
else:
  y = x
  print(y)
````

````python
# Defino una funcion para hallar el valor absoluto de un número
def absolute(x):

  if x<0:
    y = -x

  else:
    y = x

  return y

absolute(-9)
absolute(-90)
````

# 8. Repeticiones con `while`

Los métodos numéricos iterativos repiten un procedimiento hasta alcanzar una condición de parada.

`while` significa:

> repetir mientras una condición sea verdadera.


````python
iteracion = 1

while iteracion <= 5:
    print(iteracion)
    iteracion = iteracion + 1
````

## `while True` y `break`

Cuando no sabemos de antemano cuántas iteraciones serán necesarias, podemos utilizar:

```python
while True:
```

El ciclo continúa hasta encontrar un `break`.


````python
iteracion = 1

while True:
    print(iteracion)

    if iteracion == 5:
        break

    iteracion += 1
````

Esta estructura es especialmente útil en métodos numéricos porque el número de iteraciones depende de la **tolerancia** elegida.


# 9. Error aproximado y criterio de parada

En un método iterativo necesitamos comparar una aproximación actual con la anterior.

Una forma habitual de calcular el error porcentual aproximado es:


$$
\left|\varepsilon_p\right|
=
\left|
\frac{x_r^{(nuevo)}-x_r^{(anterior)}}{x_r^{(nuevo)}}
\right|100
$$



````python
x_anterior = 0.75
xr = 0.875

ep = abs((xr - x_anterior) / xr) * 100

print(ep)
````

La función incorporada `abs()` devuelve el valor absoluto.


## Tolerancia

El proceso puede detenerse cuando el error es menor que la tolerancia especificada.


````python
tolerancia = 1.0
ep = 0.8

if ep < tolerancia:
    print("Se alcanzó la tolerancia")
````

# 10. `None`: todavía no existe una aproximación anterior

En la primera iteración de bisección todavía no existe un valor anterior con el cual calcular el error.

Para representarlo podemos utilizar:


````python
x_old = None
print(x_old)
````

Después podemos preguntar:


````python
x_old = None

if x_old is not None:
    print("Ya existe una aproximación anterior")
else:
    print("Primera iteración")
````

En la primera iteración también puede ser útil asignar al error un valor inicialmente muy grande:


````python
ep = float("inf")
print(ep)
````

`float("inf")` representa infinito positivo. De esta forma, la condición `ep < tolerancia` no se cumplirá accidentalmente en la primera iteración.


# 11. Pasar una función como argumento

Este concepto permite construir un algoritmo de bisección que funcione con distintas ecuaciones.

Primero definimos una función matemática:


````python
def f1(x):
    return x**2 - 2
````

Luego podemos crear otra función que reciba `f1` como parámetro:


````python
def evaluar(func, x):
    return func(x)

print(evaluar(f1, 1.5))
````

Observá que escribimos:

```python
evaluar(f1, 1.5)
```

y no:

```python
evaluar(f1(1.5), 1.5)
```

`f1` se pasa como una **función** para que el algoritmo pueda evaluarla cuando lo necesite.

Esta misma idea se utilizará en bisección, punto fijo y Newton-Raphson.
